

> ## <mark>The result above shows many missing data </mark>



###  To fix this issue You will fill these values with mean for numeric and mode for categorical

```python
# trying to fill missing values in my datset using for loop 
num_cols = rr.select_dtypes(include=['number']).columns
for col in num_cols:
    rr[col] = rr[col].fillna(rr[col].mean())
cat_cols = rr.select_dtypes(include=['object', 'category', 'string']).columns
for col in cat_cols:
    if col != 'member_gender':
        rr[col] = rr[col].fillna(rr[col].mode()[0])

```

```python
#check the Nulls 
rr.isnull().sum()
```

### **And with a proper category for other (member gender)**

```python
##  still got a missing value in the member_gender, so let's explore this column more 

print(rr.member_gender.head(15))
```

```python
# fill  dataset with missing values.
rr['member_gender'] = rr['member_gender'].fillna('Other')
```

```python
print(rr.member_gender.head(15))
```
