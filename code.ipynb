```python
em ="Amit_ml@gmail.edu"
```

```python
if em.count("@") != 1 or "." not in em.split("@")[1]:
    print("Invalid email")
else:
    u=em.split("@")[0]
    d=em.split("@")[1].split(".")[0]
    print(f"Username: {u}, Domain: {d}")
    if em.endswith(".edu"):
        print("Educational domain")
    elif em.endswith(".com"):
        print("Commercial domain")  
    else:
        print("Other domain")
```
