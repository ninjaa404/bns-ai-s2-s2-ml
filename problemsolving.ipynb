# problem 1

```python
text = "The greatest victory is that which requires no battle"
words = text.split()
words.reverse()
result = " ".join(words)
print(result)
```

```python
text = "The greatest victory is that which requires no battle"
words = text.split()
result = []
for i in range(len(words) - 1, -1, -1):
    result.append(words[i])

print(" ".join(result))
```
