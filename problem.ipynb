##  Roman to Integer

```python
class Solution:
    def romanToInt(self, s: str) -> int:
        rm = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000,
        }
        total =0
        n = len(s)
        for i in range(n):
            if i < n-1 and rm[s[i]]<rm[s[i+1]]:
                total -=rm[s[i]]
            else:
                total +=rm[s[i]]
        return total    

```

## Search Insert Position
## بدور على مكان التارجت في التيست لو لقيته بطلع قيمه الاندكس بتاعه ملقتوش بشوف هو المفروض يبقى فين وهطلعه

```python
class Solution:
    def searchInsert(self, nums: list[int], target: int) -> int:
        for i in range(len(nums)):
            if nums[i] >= target:
                return i

        return len(nums)
```

```python
solution = Solution()
print(solution.searchInsert([1, 3, 5, 6], 5))
```

##  Basic Calculator
## فكرته انه هيدخل عددبالعمليه اللي عايز يعملها وانا المفروض هطلع الناتج بس لازم اراعي ترتيب العمليات
## اقواس /قسمه |ضرب|جمع|طرح

```python
class Solution:
    def calculate(self, s: str) -> int:
        result = 0
        last_num = 0
        num = 0
        sign = '+'

        for i in range(len(s)):
            if s[i].isdigit():
                num = num * 10 + int(s[i])

            if (not s[i].isdigit() and s[i] != ' ') or i == len(s) - 1:

                if sign == '+':
                    result += last_num
                    last_num = num

                elif sign == '-':
                    result += last_num
                    last_num = -num

                elif sign == '*':
                    last_num = last_num * num

                elif sign == '/':
                    last_num = int(last_num / num)

                sign = s[i]
                num = 0

        return result + last_num
        
```

```python
solution = Solution()
print(solution.calculate("3+2*2"))
```

## Reverse Words in a String
## بيعكس ترتيب الكلام اللي داخله ولو  فيه مسافات كتيره بيشيلها وبيخلي واخده بس 

```python
class Solution:
    def reverseWords(self, s: str) -> str:
        words = s.split()
        words.reverse()
        return " ".join(words)
```

```python
solution = Solution()
print(solution.reverseWords("  hello   world  "))
```

## Product of Array Except Self
## الفكره عندي ليستخ ارقام هطلع في كل انديكس ناتج ضرب الارقام كلها في بعض ما عدا هو

```python
class Solution:
    def productExceptSelf(self, nums: list[int]) -> list[int]:
        answer = [1] * len(nums)
        left = 1
        for i in range(len(nums)):
            answer[i] = left
            left *= nums[i]
        right = 1
        for i in range(len(nums) - 1, -1, -1):
            answer[i] *= right
            right *= nums[i]
        return answer
```

```python
solution = Solution()
print(solution.productExceptSelf([1, 2, 3, 4]))
```

## Set Matrix Zeroes
## عايز مجرد ما اشوف صفر لازم اخلي الصف والعمود اللي الصفر فيه اصفار زييه 

```python
class Solution:
    def setZeroes(self, matrix: list[list[int]]) -> None:
        """
        Do not return anything, modify matrix in-place instead.
        """
        
        rows = set()
        cols = set()

        for i in range(len(matrix)):
            for j in range(len(matrix[0])):
                if matrix[i][j] == 0:
                    rows.add(i)
                    cols.add(j)

        for i in range(len(matrix)):
            for j in range(len(matrix[0])):
                if i in rows or j in cols:
                    matrix[i][j] = 0
```

```python
solution = Solution()
matrix = [
    [1, 1, 1],
    [1, 0, 1],
    [1, 1, 1]
]
solution.setZeroes(matrix)
print(matrix)
```
