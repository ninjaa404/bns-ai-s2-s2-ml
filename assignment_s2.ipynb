# Q11

```python
myet = {10, 3, 25, 1, 7}
ml = list(myet)
maximum = ml[0]
minimum = ml[0]
for nn in ml:
    if nn > maximum:
        maximum = nn
    if nn < minimum:
        minimum = nn
print("Max:", maximum)
print("Min:", minimum)
```

# Q17

```python
keys = ["a", "b", "c"]
result = {}
for key in reversed(keys):
    result = {key: result}
print(result)
```

```python
keys = ["a", "b", "c"]
result = {}

for i in range(len(keys) - 1, -1, -1):
    result = {keys[i]: result}

print(result)
```
